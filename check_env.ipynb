# بررسی محیط کاری (check_env)
ثبت نسخه پایتون، مسیر اجرا و نسخه کتابخانه‌های پروژه تشخیص میوه.

```python
import sys, platform
print('Python version :', sys.version)
print('Python path    :', sys.executable)
print('Platform       :', platform.platform())
```

```python
import importlib
for name in ['numpy', 'matplotlib', 'sklearn', 'PIL']:
    try:
        m = importlib.import_module(name)
        print(f'{name:12s}', getattr(m, '__version__', 'unknown'))
    except ImportError:
        print(f'{name:12s} NOT INSTALLED')
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 100)
plt.plot(x, np.sin(x))
plt.title('Matplotlib works')
plt.show()
```
